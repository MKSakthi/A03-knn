#Q1.1
Regression predicts a number whereas classification predicts a category or class. For example, predicting sales is regression but predicting a type of car is classification.

#Q1.2
A confusion table compares the actual classes to the predicted classes. It shows how many predictions were correct and what type of mistakes the model made

#Q1.3
SSE measures how far the model's predictions are from the actual values. (The lower the SSE, the better the prediction)

#Q1.4
Overfitting means the model follows the training data too closely and may not work well on new data and Underfitting means the model is too simple and doesn't capture the patterns in the data

#Q1.5
We use the training data to build the model and test data to see how well it works on new data. We can try different values of k and choose the one with the best accuracy for classification or lowest SSE for regression

#Q1.6
Class label gives one simple prediction and a probability distriction gives more information about how likely each class is. A class label is easier to understand but the probabilities are useful because they show how certain or uncertain a prediction is

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

#Q2.1
#Loading the data and EDA
df = pd.read_csv("/content/A03-knn/data/land_mines.csv")
print(df.shape)
print("")
print(df.head(5))
print("")
print(df.describe())
print("")
print(df["mine_type"].value_counts())

#Q2.2
# Features and target
X =df[["voltage", "height", "soil"]]
y =df["mine_type"]

# 50/50 training and test split
Xtrain, Xtest, ytrain, ytest = train_test_split(X,y,test_size=0.5,random_state = 42, stratify=y)

#scales from 0 to 1
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(Xtrain)
X_test_scaled = scaler.transform(Xtest)

#Q2.3
print("\nQ2.3")
# Trying different values of k
accuracies = []
for k in range (1,21):
  model=KNeighborsClassifier(n_neighbors=k)
  fitted_model = model.fit(X_train_scaled, ytrain)
  yhat=fitted_model.predict(X_test_scaled)
  accuracies.append(accuracy_score(ytest, yhat))
bestk = accuracies.index(max(accuracies))+1
print("Best k:", bestk)
print("Best accuracy:", max(accuracies))
print("I tested diffrent values for k from 1 to 20 ans selected the one with the highest accuracy")

#Q2.4
print("\nQ2.4")
#Fitting the model using the best k
model = KNeighborsClassifier(n_neighbors=bestk)
fitted_model = model.fit(X_train_scaled, ytrain)
yhat=fitted_model.predict(X_test_scaled)
#Confusion table
confusion_table = pd.crosstab(ytest,yhat,rownames=["Actual"],colnames=["Predicted"])
print(confusion_table)
print("Accuracy:", accuracy_score(ytest,yhat)) #Final accuracy
print("The model is 46.15% accurate. it performs best for mine type 2 which correctly predicts 32 out of 35 cases. It performs less accurately for mine types 3 and 5, with only 7 correct predictions for each. The model often confuses mine types 3 and 5 with other mine types.")

#Q2.5
print("\nQ2.5")
print("I would not use this model by itself to identify land mines becuse it makes many incorrect predictions. it could be used as an additional tool, but the predicted mine type should be confirmed using other methods, esepcially when the model is uncertain")
```
